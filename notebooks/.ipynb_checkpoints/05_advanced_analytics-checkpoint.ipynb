```python
import pandas as pd

df = pd.read_csv("../data/processed/nav_data.csv")

df["daily_return"] = df["nav"].pct_change()

df.head()
```

```python
var95 = df["daily_return"].quantile(0.05)

print("VaR 95%:", var95)
```
